## 3. Lihat Hasil

```python
import json
from IPython.display import Image, display

for name in ["cnn", "gru"]:
    metrics = json.load(open(f"results/{name}_metrics.json"))
    print(f"== {metrics['model']} | {metrics['dataset']} ==")
    print(f"Akurasi test: {metrics['test_accuracy']*100:.2f}% | waktu: {metrics['train_seconds']}s")
    display(Image(filename=f"results/{name}_confusion_matrix.png"))
    display(Image(filename=f"results/{name}_curves.png"))
```
